```python
input_file = "/data1/lesliec/carolw/data_raw/CNV_calls/5C54B79C_DA02_4B22_9FC2_3D61BFFC5559.csv"   # path to your CSV
baseline_ploidy = 2            # change if tumor ploidy known
high_amp_threshold = 6         # define high amplification cutoff
```

```python
df = pd.read_csv(input_file)

# Ensure numeric types
df["Start"] = pd.to_numeric(df["Start"])
df["End"] = pd.to_numeric(df["End"])
df["Copy_Number"] = pd.to_numeric(df["Copy_Number"])
df["Major_Copy_Number"] = pd.to_numeric(df["Major_Copy_Number"])
df["Minor_Copy_Number"] = pd.to_numeric(df["Minor_Copy_Number"])
```

```python
df["Length"] = df["End"] - df["Start"]
total_genome_length = df["Length"].sum()
```

```python
# fraction of genome altered
altered_length = df.loc[df["Copy_Number"] != baseline_ploidy, "Length"].sum()
fraction_genome_altered = altered_length / total_genome_length
```

```python
# length weighted mean absolute CN deviation
df["CN_deviation"] = (df["Copy_Number"] - baseline_ploidy).abs()
mean_abs_deviation = (df["CN_deviation"] * df["Length"]).sum() / total_genome_length
```

```python
# High amplification burden
high_amp_length = df.loc[df["Copy_Number"] >= high_amp_threshold, "Length"].sum()
fraction_high_amp = high_amp_length / total_genome_length
```

```python
print("CNV Burden Metrics")
print("-------------------")
print(f"Total genome length analyzed: {total_genome_length:,} bp")
print(f"Fraction of genome altered: {fraction_genome_altered:.4f}")
print(f"Mean absolute CN deviation: {mean_abs_deviation:.4f}")
print(f"Fraction of high amplification (CN ≥ {high_amp_threshold}): {fraction_high_amp:.4f}")
```
